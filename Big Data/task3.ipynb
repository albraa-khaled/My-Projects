```python
from pyspark import SparkContext, SparkConf
import findspark
```

```python
findspark.init()
```

```python
conf = SparkConf().setMaster("local").setAppName("my project")
```

```python
sc = SparkContext(conf=conf)
```

```python
file_path = "C:/Users/User/spark/in/trip.txt"
rdd_base = sc.textFile(file_path, 3)
```

```python
rdd_split1=rdd_base.map(lambda x:x.split(","))
```

```python
rdd_filter1=rdd_split1.filter(lambda x: x[6]=="Air France")
```

```python
rdd_filter2=rdd_filter1.filter(lambda x: x[3]=="Kuala Lumpur International Airport")
```

```python
rdd_filter2.saveAsTextFile('Air France_Kuala Lumpur International Airport.txt')
```
