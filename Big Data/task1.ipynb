```python
from pyspark import SparkContext, SparkConf
import findspark
```

```python
findspark.init()
```

```python
conf = SparkConf().setMaster("local").setAppName("my project")
conf.set("spark.network.timeout", "600s")
```

```python
sc = SparkContext(conf=conf)
```

```python
file_path = "C:/Users/User/spark/in/trip.txt"
rdd_base = sc.textFile(file_path, 3)
```

```python
rdd_split1=rdd_base.map(lambda x:x.split(","))
```

```python
rdd_split1.take(4)
```

```python
rdd_split2=rdd_split1.filter(lambda x: x[1]=="McCarran International Airport")
```

```python
rdd_split2.take(4)
```

```python
PairRdd=rdd_split2.map(lambda x : (x[0][:7],1))
```

```python
PairRdd.take(10)
```

```python
PairRdd.reduceByKey(lambda x,y:x+y).collect()
```
