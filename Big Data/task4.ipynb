```python
from pyspark import SparkContext, SparkConf
import findspark
```

```python
findspark.init()
```

```python
conf = SparkConf().setMaster("local").setAppName("my project")
conf.set("spark.network.timeout", "600s")
```

```python
sc = SparkContext(conf=conf)
```

```python
file_path = "C:/Users/User/spark/in/trip.txt"
rdd_base = sc.textFile(file_path, 3)
```

```python
rdd_split1=rdd_base.map(lambda x:x.split(","))
```

```python
from datetime import datetime, timedelta
```

```python
start_time1 = datetime.strptime('20:00', '%H:%M').time()
end_time1 = datetime.strptime('23:00', '%H:%M').time()
final_time = datetime.strptime('03:00', '%H:%M').time()
```

```python
rdd_filter2 = rdd_split1.filter(lambda x:datetime.strptime(x[0], '%Y-%m-%d %H:%M:%S').date() < datetime.strptime(x[5], '%Y-%m-%d %H:%M:%S').date() )
```

```python
rdd_filter3 = rdd_filter2.filter(lambda x: start_time1 <= datetime.strptime(x[0], '%Y-%m-%d %H:%M:%S').time() <= end_time1 )
```

```python
rdd_airlines=rdd_filter3.map(lambda x : (x[6],1))
```

```python
rdd_airlines.take(5)
```

```python
rdd_airlines1=rdd_airlines.reduceByKey(lambda x,y:x+y)
```
