```python
from pyspark import SparkContext, SparkConf
import findspark
```

```python
findspark.init()
```

```python
conf = SparkConf().setMaster("local").setAppName("my project")
```

```python
sc = SparkContext(conf=conf)
```

```python
file_path = "C:/Users/User/spark/in/trip.txt"
rdd_base = sc.textFile(file_path, 3)
```

```python
rdd_split1=rdd_base.map(lambda x:x.split(","))
```

```python
rdd_filter1=rdd_split1.filter(lambda x: x[1]=="Frankfurt Airport")
```

```python
from datetime import datetime, timedelta
```

```python
start_time = datetime.strptime('13:00', '%H:%M').time()
end_time = datetime.strptime('22:00', '%H:%M').time()
```

```python
rdd_filter2 = rdd_filter1.filter(lambda x: start_time <= datetime.strptime(x[0], '%Y-%m-%d %H:%M:%S').time() <= end_time)
```

```python
rdd_map1=rdd_filter2.map(lambda x : (x[0][11:16],int(x[10])))
```

```python
values_rdd1 = rdd_map1.map(lambda x: x[1])
```

```python
sum_rdd1= values_rdd1.reduce(lambda x, y: x + y)
print("the number of passengers departing from “Frankfurt Airport” between 13:00 and 22:00 = ",sum_rdd1,"passenger")
```
