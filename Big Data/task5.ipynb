```python
from pyspark import SparkContext, SparkConf
import findspark
```

```python
findspark.init()
```

```python
conf = SparkConf().setMaster("local").setAppName("my project")
```

```python
sc = SparkContext(conf=conf)
```

```python
file_path = "C:/Users/User/spark/in/trip.txt"
rdd_base = sc.textFile(file_path, 3)
```

```python
rdd_split1=rdd_base.map(lambda x:x.split(","))
```

```python
rdd_task5=rdd_split1.filter(lambda x: x[6]=="Air Canada")
```

```python
from datetime import datetime, timedelta
```

```python
date=datetime.strptime("2024-05",'%Y-%m').date()
```

```python
rdd_task51 = rdd_task5.filter(lambda x: date==datetime.strptime(x[0], '%Y-%m-%d %H:%M:%S').date())
```

```python
rdd_task51.take(20)
```

```python
PairRdd3 = rdd_task51.map(lambda x: (x,str(int(((float(x[10]) / float(x[9])) * 100))) ))
```

```python
PairRdd3.take(10)
```
